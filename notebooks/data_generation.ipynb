```python
# -------------------------
# CONFIGURATION
# -------------------------
num_records = 100000
file_name = "mazra3ti_data_test.csv"
np.random.seed(42)

# 35 Crops with environmental and economic preferences
crops_info = {
    'Wheat': {'cat': 'Cereal', 'sea': 'Winter', 'yield_range': (0.5, 0.8), 'base_price': 0.45},
    'Barley': {'cat': 'Cereal', 'sea': 'Winter', 'yield_range': (0.4, 0.7), 'base_price': 0.35},
    'Corn': {'cat': 'Cereal', 'sea': 'Summer', 'yield_range': (0.6, 1.0), 'base_price': 0.55},
    'Rice': {'cat': 'Cereal', 'sea': 'Summer', 'yield_range': (0.7, 1.1), 'base_price': 0.65},

    'Tomato': {'cat': 'Vegetable', 'sea': 'Spring', 'yield_range': (2.0, 5.0), 'base_price': 1.2},
    'Potato': {'cat': 'Vegetable', 'sea': 'Fall', 'yield_range': (1.0, 3.0), 'base_price': 0.7},
    'Cucumber': {'cat': 'Vegetable', 'sea': 'Spring', 'yield_range': (1.0, 3.0), 'base_price': 0.9},
    'Eggplant': {'cat': 'Vegetable', 'sea': 'Summer', 'yield_range': (1.5, 3.5), 'base_price': 0.8},
    'Bell Pepper': {'cat': 'Vegetable', 'sea': 'Summer', 'yield_range': (1.0, 2.5), 'base_price': 1.4},
    'Onion': {'cat': 'Vegetable', 'sea': 'Fall', 'yield_range': (2.0, 4.0), 'base_price': 0.5},
    'Garlic': {'cat': 'Vegetable', 'sea': 'Winter', 'yield_range': (0.8, 1.5), 'base_price': 2.5},
    'Lettuce': {'cat': 'Vegetable', 'sea': 'Winter', 'yield_range': (1.0, 2.0), 'base_price': 0.6},
    'Spinach': {'cat': 'Vegetable', 'sea': 'Winter', 'yield_range': (0.5, 1.2), 'base_price': 0.9},
    'Carrot': {'cat': 'Vegetable', 'sea': 'Winter', 'yield_range': (1.0, 2.5), 'base_price': 0.7},
    'Zucchini': {'cat': 'Vegetable', 'sea': 'Spring', 'yield_range': (1.5, 3.0), 'base_price': 0.8},
    'Cabbage': {'cat': 'Vegetable', 'sea': 'Winter', 'yield_range': (2.0, 4.0), 'base_price': 0.6},
    'Broccoli': {'cat': 'Vegetable', 'sea': 'Winter', 'yield_range': (1.0, 2.2), 'base_price': 1.3},

    'Chickpeas': {'cat': 'Legume', 'sea': 'Spring', 'yield_range': (0.2, 0.5), 'base_price': 1.8},
    'Lentils': {'cat': 'Legume', 'sea': 'Winter', 'yield_range': (0.15, 0.4), 'base_price': 1.5},
    'Fava Beans': {'cat': 'Legume', 'sea': 'Winter', 'yield_range': (0.4, 0.9), 'base_price': 1.1},
    'Beans': {'cat': 'Legume', 'sea': 'Spring', 'yield_range': (0.5, 1.2), 'base_price': 1.4},
    'Peas': {'cat': 'Legume', 'sea': 'Winter', 'yield_range': (0.4, 1.0), 'base_price': 1.3},

    'Strawberry': {'cat': 'Fruit', 'sea': 'Spring', 'yield_range': (1.0, 2.5), 'base_price': 3.5},
    'Watermelon': {'cat': 'Fruit', 'sea': 'Summer', 'yield_range': (3.0, 6.0), 'base_price': 0.6},
    'Melon': {'cat': 'Fruit', 'sea': 'Summer', 'yield_range': (2.0, 4.5), 'base_price': 0.8},
    'Grapes': {'cat': 'Fruit', 'sea': 'Summer', 'yield_range': (1.5, 3.5), 'base_price': 1.8},
    'Olives': {'cat': 'Fruit', 'sea': 'Fall', 'yield_range': (1.5, 3.0), 'base_price': 2.5},
    'Apple': {'cat': 'Fruit', 'sea': 'Fall', 'yield_range': (1.0, 3.0), 'base_price': 2.0},
    'Peach': {'cat': 'Fruit', 'sea': 'Spring', 'yield_range': (0.8, 2.0), 'base_price': 2.2},
    'Apricot': {'cat': 'Fruit', 'sea': 'Spring', 'yield_range': (0.7, 1.8), 'base_price': 2.4},
    'Pomegranate': {'cat': 'Fruit', 'sea': 'Fall', 'yield_range': (1.2, 2.8), 'base_price': 1.7},
    'Dates': {'cat': 'Fruit', 'sea': 'Summer', 'yield_range': (0.5, 2.0), 'base_price': 4.5},
    'Orange': {'cat': 'Fruit', 'sea': 'Winter', 'yield_range': (2.0, 4.5), 'base_price': 1.6},
    'Lemon': {'cat': 'Fruit', 'sea': 'Winter', 'yield_range': (1.8, 4.0), 'base_price': 1.5},
    'Banana': {'cat': 'Fruit', 'sea': 'Summer', 'yield_range': (2.5, 5.0), 'base_price': 2.2},
}

locations = ['Amman', 'Irbid', 'Aqaba', 'Zarqa', 'Madaba', 'Mafraq', 'Jerash', 'Ajloun', 'Karak', 'Tafilah', 'Maan', 'Balqa']
region_types = ['Urban', 'Rural', 'Valley', 'Desert']
customer_types = ['Trader', 'Local Market', 'Export']

print(f"--- Starting generation of {num_records} records ---")

# -------------------------
# 1. Categories and Basic Info
# -------------------------
record_ids = np.arange(1, num_records + 1)
chosen_crops = np.random.choice(list(crops_info.keys()), num_records)
r_types = np.random.choice(region_types, num_records, p=[0.12, 0.38, 0.28, 0.22])
f_locs = np.random.choice(locations, num_records)
m_locs = np.random.choice(locations, num_records)
cust_types = np.random.choice(customer_types, num_records, p=[0.35, 0.45, 0.20])

# -------------------------
# 2. Environmental Factors
# -------------------------
temp_map = {'Desert': 35, 'Valley': 30, 'Rural': 23, 'Urban': 25}
rain_scale_map = {'Desert': 3.5, 'Valley': 6.5, 'Rural': 8.5, 'Urban': 5.5}
humidity_range_map = {
    'Desert': (18, 45),
    'Valley': (35, 65),
    'Rural': (30, 70),
    'Urban': (25, 60)
}
wind_range_map = {
    'Desert': (5, 28),
    'Valley': (3, 18),
    'Rural': (2, 16),
    'Urban': (2, 14)
}

temp_base = np.array([temp_map[r] for r in r_types])
temperature = np.random.normal(temp_base, 3.5, num_records)
rainfall = np.array([np.random.gamma(shape=2.2, scale=rain_scale_map[r]) for r in r_types])
humidity = np.array([np.random.uniform(*humidity_range_map[r]) for r in r_types])
wind_speed = np.array([np.random.uniform(*wind_range_map[r]) for r in r_types])

temperature = np.clip(temperature, 5, 50)
rainfall = np.clip(rainfall, 0.1, 120)
humidity = np.clip(humidity, 18, 90)
wind_speed = np.clip(wind_speed, 2, 30)

# -------------------------
# 3. Dates Logic with season-aware planting
# -------------------------
season_months = {
    'Winter': [11, 12, 1, 2],
    'Spring': [3, 4, 5],
    'Summer': [6, 7, 8],
    'Fall': [9, 10]
}

planting_dates = []
for crop in chosen_crops:
    crop_season = crops_info[crop]['sea']
    possible_months = season_months[crop_season]
    year = np.random.randint(2020, 2026)
    month = np.random.choice(possible_months)
    day = np.random.randint(1, 29)
    planting_dates.append(np.datetime64(f"{year:04d}-{month:02d}-{day:02d}"))
planting_dates = np.array(planting_dates, dtype='datetime64[D]')

crop_duration_map = {
    'Wheat': (120, 170), 'Barley': (110, 160), 'Corn': (90, 140), 'Rice': (100, 150),
    'Tomato': (75, 120), 'Potato': (90, 130), 'Cucumber': (65, 100),
    'Eggplant': (90, 135), 'Bell Pepper': (95, 140), 'Onion': (100, 140),
    'Garlic': (110, 160), 'Lettuce': (55, 85), 'Spinach': (45, 75),
    'Carrot': (75, 120), 'Zucchini': (60, 95), 'Cabbage': (90, 140), 'Broccoli': (75, 120),
    'Chickpeas': (90, 130), 'Lentils': (95, 135), 'Fava Beans': (95, 140),
    'Beans': (70, 110), 'Peas': (65, 105),
    'Strawberry': (100, 150), 'Watermelon': (80, 115), 'Melon': (75, 110),
    'Grapes': (110, 160), 'Olives': (140, 190), 'Apple': (120, 170),
    'Peach': (100, 145), 'Apricot': (95, 140), 'Pomegranate': (115, 165),
    'Dates': (130, 180), 'Orange': (120, 180), 'Lemon': (110, 170), 'Banana': (140, 210)
}

harvest_days = np.array([
    np.random.randint(crop_duration_map[c][0], crop_duration_map[c][1] + 1)
    for c in chosen_crops
])

harvest_dates = planting_dates + harvest_days.astype('timedelta64[D]')
sale_gap_days = np.random.randint(1, 31, num_records)
sale_dates = harvest_dates + sale_gap_days.astype('timedelta64[D]')

# -------------------------
# 4. Yield & Financials
# -------------------------
area_size = np.random.uniform(0.5, 120, num_records)

min_y = np.array([crops_info[c]['yield_range'][0] for c in chosen_crops])
max_y = np.array([crops_info[c]['yield_range'][1] for c in chosen_crops])

region_yield_factor = np.select(
    [r_types == 'Valley', r_types == 'Rural', r_types == 'Urban', r_types == 'Desert'],
    [1.20, 1.05, 0.95, 0.78],
    default=1.0
)

heat_stress = np.where(temperature > 37, 0.85, 1.0)
wind_stress = np.where(wind_speed > 22, 0.92, 1.0)
rain_bonus = np.where((rainfall >= 8) & (rainfall <= 25), 1.08, 1.0)

pest_probability = (
    0.02
    + np.where(humidity > 60, 0.05, 0.0)
    + np.where(rainfall > 25, 0.04, 0.0)
    + np.where(temperature > 34, 0.03, 0.0)
)
pest_probability = np.clip(pest_probability, 0.02, 0.20)
pest_indicator = np.random.random(num_records) < pest_probability

potential_yield = (
    area_size
    * np.random.uniform(min_y * 1.05, max_y * 1.45)
    * region_yield_factor
    * rain_bonus
)

actual_yield = (
    potential_yield
    * heat_stress
    * wind_stress
    * np.where(
        pest_indicator,
        np.random.uniform(0.74, 0.91, num_records),
        np.random.uniform(0.94, 1.04, num_records)
    )
)
actual_yield = np.maximum(actual_yield, 0.05)

water_used = (
    area_size
    * np.random.uniform(6, 16, num_records)
    * np.where(
        r_types == 'Desert', 1.35,
        np.where(r_types == 'Valley', 1.10,
                 np.where(r_types == 'Urban', 1.00, 0.90))
    )
)

fertilizer_used = area_size * np.random.uniform(0.20, 0.70, num_records)

sell_ratio = np.select(
    [cust_types == 'Export', cust_types == 'Trader', cust_types == 'Local Market'],
    [
        np.random.uniform(0.945, 0.995, num_records),
        np.random.uniform(0.89, 0.975, num_records),
        np.random.uniform(0.84, 0.955, num_records)
    ]
)

quantity_sold = actual_yield * sell_ratio
inventory_remaining = np.maximum(actual_yield - quantity_sold, 0)

base_prices = np.array([crops_info[c]['base_price'] for c in chosen_crops])

region_price_factor = np.select(
    [m_locs == 'Amman', m_locs == 'Aqaba', m_locs == 'Irbid'],
    [1.10, 1.07, 1.05],
    default=1.0
)

customer_price_factor = np.select(
    [cust_types == 'Export', cust_types == 'Trader', cust_types == 'Local Market'],
    [1.22, 1.03, 1.06],
    default=1.0
)

seasonal_noise = np.random.uniform(1.00, 1.15, num_records)

unit_price = (
    base_prices
    * region_price_factor
    * customer_price_factor
    * seasonal_noise
    * 1.52
)
unit_price = np.minimum(unit_price, 6.5)

market_price = unit_price * np.random.uniform(0.97, 1.08, num_records)
total_sales = quantity_sold * unit_price

dist_to_market = np.where(
    f_locs == m_locs,
    np.random.uniform(3, 20, num_records),
    np.random.uniform(25, 260, num_records)
)

transport_cost = (
    dist_to_market * np.random.uniform(0.007, 0.025, num_records)
    + quantity_sold * np.random.uniform(0.0025, 0.010, num_records)
)

land_preparation_cost = area_size * np.random.uniform(0.55, 1.50, num_records)
irrigation_cost = water_used * np.random.uniform(0.004, 0.011, num_records)
fertilizer_cost = fertilizer_used * np.random.uniform(0.14, 0.30, num_records)
labor_cost = area_size * np.random.uniform(0.32, 0.95, num_records)
pest_control_cost = np.where(
    pest_indicator,
    area_size * np.random.uniform(0.10, 0.35, num_records),
    area_size * np.random.uniform(0.02, 0.10, num_records)
)
packaging_cost = quantity_sold * np.random.uniform(0.004, 0.014, num_records)

total_expenses = (
    land_preparation_cost
    + irrigation_cost
    + fertilizer_cost
    + labor_cost
    + pest_control_cost
    + packaging_cost
    + transport_cost
)

profit = total_sales - total_expenses

# -------------------------
# 5. Demand, Supply, Sufficiency Features
# -------------------------
def minmax_norm(x):
    return (x - x.min()) / (x.max() - x.min() + 1e-6)

# ---- Demand score ----
sales_ratio = np.clip(quantity_sold / np.maximum(actual_yield, 1e-6), 0, 1.0)
crop_avg_market_price = pd.Series(market_price).groupby(chosen_crops).transform('mean').values
relative_price_index = market_price / np.where(crop_avg_market_price == 0, 1, crop_avg_market_price)
sale_speed_score = 1 - np.clip((sale_gap_days - 1) / 30, 0, 1)

customer_demand_weight = np.select(
    [cust_types == 'Export', cust_types == 'Trader', cust_types == 'Local Market'],
    [1.18, 1.00, 0.93],
    default=1.0
)

market_demand_weight = np.select(
    [m_locs == 'Amman', m_locs == 'Aqaba', m_locs == 'Irbid'],
    [1.15, 1.08, 1.05],
    default=1.0
)

crop_seasons = np.array([crops_info[c]['sea'] for c in chosen_crops])
season_demand_weight = np.select(
    [crop_seasons == 'Summer', crop_seasons == 'Spring', crop_seasons == 'Winter', crop_seasons == 'Fall'],
    [1.05, 1.02, 0.98, 0.97],
    default=1.0
)

inventory_pressure = 1 - np.clip(inventory_remaining / np.maximum(actual_yield, 1e-6), 0, 1)

demand_score_raw = (
    0.30 * sales_ratio +
    0.18 * relative_price_index +
    0.16 * sale_speed_score +
    0.14 * inventory_pressure +
    0.12 * customer_demand_weight +
    0.07 * market_demand_weight +
    0.03 * season_demand_weight
)

demand_score = minmax_norm(demand_score_raw)

demand_level = np.select(
    [
        demand_score < 0.36,
        (demand_score >= 0.36) & (demand_score < 0.64),
        demand_score >= 0.64
    ],
    ['Low', 'Medium', 'High'],
    default='Medium'
)

# ---- Supply score ----
yield_per_area = actual_yield / np.maximum(area_size, 1e-6)
water_efficiency = actual_yield / np.maximum(water_used, 1e-6)
fertilizer_efficiency = actual_yield / np.maximum(fertilizer_used, 1e-6)

climate_penalty = (
    (temperature > 37).astype(int) +
    (wind_speed > 22).astype(int) +
    (rainfall > 25).astype(int) +
    pest_indicator.astype(int)
)

yield_per_area_norm = minmax_norm(yield_per_area)
area_size_norm = minmax_norm(area_size)
water_efficiency_norm = minmax_norm(water_efficiency)
fertilizer_efficiency_norm = minmax_norm(fertilizer_efficiency)
actual_yield_norm = minmax_norm(actual_yield)
climate_penalty_norm = climate_penalty / np.maximum(climate_penalty.max(), 1)

supply_score_raw = (
    0.24 * yield_per_area_norm +
    0.16 * area_size_norm +
    0.22 * water_efficiency_norm +
    0.20 * fertilizer_efficiency_norm +
    0.10 * actual_yield_norm +
    0.08 * (1 - climate_penalty_norm)
)

supply_score = minmax_norm(supply_score_raw)

supply_level = np.select(
    [
        supply_score < 0.28,
        (supply_score >= 0.28) & (supply_score < 0.50),
        supply_score >= 0.50
    ],
    ['Low', 'Medium', 'High'],
    default='Medium'
)

# ---- Sufficiency ----
# Use a slightly centered ratio so Balanced is not too rare
adjusted_supply_score = np.clip(0.92 * supply_score + 0.08 * demand_score, 0, 1)
supply_demand_ratio = np.round(adjusted_supply_score / np.maximum(demand_score, 1e-4), 6)

sufficiency_status = np.select(
    [
        supply_demand_ratio < 0.92,
        (supply_demand_ratio >= 0.92) & (supply_demand_ratio <= 1.08),
        supply_demand_ratio > 1.08
    ],
    ['Shortage', 'Balanced', 'Surplus'],
    default='Balanced'
)

# -------------------------
# 6. DataFrame Construction
# -------------------------
df = pd.DataFrame({
    'record_id': record_ids,
    'crop_name': chosen_crops,
    'planting_date': planting_dates,
    'harvest_date': harvest_dates,
    'season': crop_seasons,
    'area_size': np.round(area_size, 2),

    'quantity_sold': np.round(quantity_sold, 2),
    'inventory_remaining': np.round(inventory_remaining, 2),

    'unit_price': np.round(unit_price, 3),
    'total_sales': np.round(total_sales, 2),
    'total_expenses': np.round(total_expenses, 2),
    'fertilizer_used': np.round(fertilizer_used, 2),
    'water_used': np.round(water_used, 2),
    'actual_yield': np.round(actual_yield, 2),
    'profit': np.round(profit, 2),
    'sale_date': sale_dates,
    'sale_gap_days': sale_gap_days,
    'market_location': m_locs,
    'farm_location': f_locs,
    'region_type': r_types,
    'distance_to_market': np.round(dist_to_market, 2),
    'transport_cost': np.round(transport_cost, 2),
    'customer_type': cust_types,
    'market_price': np.round(market_price, 3),
    'temperature': np.round(temperature, 1),
    'humidity': np.round(humidity, 1),
    'rainfall': np.round(rainfall, 1),
    'wind_speed': np.round(wind_speed, 1),
    'pest_indicator': pest_indicator.astype(bool),

    'water_efficiency': np.round(water_efficiency, 4),
    'fertilizer_efficiency': np.round(fertilizer_efficiency, 4),

    'demand_score': np.round(demand_score, 4),
    'demand_level': demand_level.astype(str),
    'supply_score': np.round(supply_score, 4),
    'supply_level': supply_level.astype(str),
    'sufficiency_status': sufficiency_status
})

print(f"--- Finished generating {num_records} records ---")
```

```python
def check_dataset_realism_advanced(df):
    import numpy as np
    import pandas as pd

    print("\n==========================================")
    print(" ADVANCED DATASET REALISM CHECK REPORT ")
    print("==========================================")

    report = {}

    # -------------------------------
    # Helper
    # -------------------------------
    def safe_pct(mask):
        return round(float(mask.mean() * 100), 4) if len(mask) > 0 else 0.0

    # ==================================================
    # 0. BASIC DATASET OVERVIEW
    # ==================================================
    print("\n[0] BASIC DATASET OVERVIEW")

    report["row_count"] = len(df)
    report["column_count"] = df.shape[1]
    report["missing_values_total"] = int(df.isnull().sum().sum())
    report["duplicate_rows"] = int(df.duplicated().sum())

    print("Rows:", report["row_count"])
    print("Columns:", report["column_count"])
    print("Total missing values:", report["missing_values_total"])
    print("Duplicate rows:", report["duplicate_rows"])

    # ==================================================
    # 1. RECORD ID CHECK
    # ==================================================
    print("\n[1] RECORD ID CHECK")

    report["record_id_unique"] = bool(df["record_id"].is_unique)
    report["record_id_duplicates"] = int(df["record_id"].duplicated().sum())
    report["record_id_min"] = int(df["record_id"].min())
    report["record_id_max"] = int(df["record_id"].max())
    report["record_id_missing_sequence_count"] = int(
        len(set(range(int(df["record_id"].min()), int(df["record_id"].max()) + 1)) - set(df["record_id"]))
    )

    print("Unique IDs:", report["record_id_unique"])
    print("Duplicate IDs:", report["record_id_duplicates"])
    print("Min ID:", report["record_id_min"])
    print("Max ID:", report["record_id_max"])
    print("Missing IDs in sequence:", report["record_id_missing_sequence_count"])

    # ==================================================
    # 2. CATEGORICAL VALIDATION
    # ==================================================
    print("\n[2] CATEGORICAL VALIDATION")

    allowed_crops = set(crops_info.keys())
    allowed_seasons = {"Winter", "Spring", "Summer", "Fall"}
    allowed_locations = set(locations)
    allowed_region_types = set(region_types)
    allowed_customer_types = set(customer_types)
    allowed_levels = {"Low", "Medium", "High"}
    allowed_sufficiency = {"Shortage", "Balanced", "Surplus"}

    report["invalid_crop_name_count"] = int((~df["crop_name"].isin(allowed_crops)).sum())
    report["invalid_season_count"] = int((~df["season"].isin(allowed_seasons)).sum())
    report["invalid_market_location_count"] = int((~df["market_location"].isin(allowed_locations)).sum())
    report["invalid_farm_location_count"] = int((~df["farm_location"].isin(allowed_locations)).sum())
    report["invalid_region_type_count"] = int((~df["region_type"].isin(allowed_region_types)).sum())
    report["invalid_customer_type_count"] = int((~df["customer_type"].isin(allowed_customer_types)).sum())
    report["invalid_demand_level_count"] = int((~df["demand_level"].isin(allowed_levels)).sum())
    report["invalid_supply_level_count"] = int((~df["supply_level"].isin(allowed_levels)).sum())
    report["invalid_sufficiency_status_count"] = int((~df["sufficiency_status"].isin(allowed_sufficiency)).sum())

    print("Invalid crop names:", report["invalid_crop_name_count"])
    print("Invalid seasons:", report["invalid_season_count"])
    print("Invalid market locations:", report["invalid_market_location_count"])
    print("Invalid farm locations:", report["invalid_farm_location_count"])
    print("Invalid region types:", report["invalid_region_type_count"])
    print("Invalid customer types:", report["invalid_customer_type_count"])
    print("Invalid demand levels:", report["invalid_demand_level_count"])
    print("Invalid supply levels:", report["invalid_supply_level_count"])
    print("Invalid sufficiency status:", report["invalid_sufficiency_status_count"])

    # ==================================================
    # 3. REQUIRED COLUMNS CHECK
    # ==================================================
    print("\n[3] REQUIRED COLUMNS CHECK")

    required_cols = [
        "record_id", "crop_name", "planting_date", "harvest_date", "sale_date", "season",
        "area_size", "quantity_sold", "unit_price", "total_sales", "total_expenses",
        "fertilizer_used", "water_used", "actual_yield", "profit",
        "market_location", "farm_location", "region_type", "distance_to_market",
        "transport_cost", "customer_type", "market_price",
        "temperature", "humidity", "rainfall", "wind_speed", "pest_indicator",
        "demand_score", "demand_level", "supply_score", "supply_level", "sufficiency_status"
    ]

    missing_required = [c for c in required_cols if c not in df.columns]
    report["missing_required_columns"] = missing_required
    report["missing_required_columns_count"] = len(missing_required)

    print("Missing required columns:", missing_required)

    # ==================================================
    # 4. CROP-SEASON CONSISTENCY
    # ==================================================
    print("\n[4] CROP-SEASON CONSISTENCY")

    crop_season_match = df.apply(
        lambda x: x["season"] == crops_info[x["crop_name"]]["sea"] if x["crop_name"] in crops_info else False,
        axis=1
    )
    report["crop_season_mismatch_count"] = int((~crop_season_match).sum())

    print("Crop-season mismatches:", report["crop_season_mismatch_count"])

    # ==================================================
    # 5. DATE LOGIC
    # ==================================================
    print("\n[5] DATE LOGIC")

    df_temp = df.copy()
    df_temp["planting_date"] = pd.to_datetime(df_temp["planting_date"])
    df_temp["harvest_date"] = pd.to_datetime(df_temp["harvest_date"])
    df_temp["sale_date"] = pd.to_datetime(df_temp["sale_date"])

    crop_duration_days = (df_temp["harvest_date"] - df_temp["planting_date"]).dt.days
    sale_gap_days = (df_temp["sale_date"] - df_temp["harvest_date"]).dt.days

    report["bad_date_order_count"] = int(
        ((df_temp["planting_date"] >= df_temp["harvest_date"]) |
         (df_temp["harvest_date"] >= df_temp["sale_date"])).sum()
    )

    report["crop_duration_too_short_count"] = int((crop_duration_days < 30).sum())
    report["crop_duration_too_long_count"] = int((crop_duration_days > 250).sum())
    report["sale_gap_too_short_count"] = int((sale_gap_days < 1).sum())
    report["sale_gap_too_long_count"] = int((sale_gap_days > 45).sum())

    print("Wrong planting → harvest → sale order:", report["bad_date_order_count"])
    print("Crop duration too short:", report["crop_duration_too_short_count"])
    print("Crop duration too long:", report["crop_duration_too_long_count"])
    print("Sale gap too short:", report["sale_gap_too_short_count"])
    print("Sale gap too long:", report["sale_gap_too_long_count"])

    # ==================================================
    # 6. CROP-SPECIFIC DURATION CHECK
    # ==================================================
    print("\n[6] CROP-SPECIFIC DURATION CHECK")

    expected_min = df_temp["crop_name"].map(lambda c: crop_duration_map[c][0])
    expected_max = df_temp["crop_name"].map(lambda c: crop_duration_map[c][1])

    report["crop_duration_below_crop_min_count"] = int((crop_duration_days < expected_min).sum())
    report["crop_duration_above_crop_max_count"] = int((crop_duration_days > expected_max).sum())

    print("Crop duration below expected crop minimum:", report["crop_duration_below_crop_min_count"])
    print("Crop duration above expected crop maximum:", report["crop_duration_above_crop_max_count"])

    # ==================================================
    # 7. PLANTING MONTH CONSISTENCY
    # ==================================================
    print("\n[7] PLANTING MONTH CONSISTENCY")

    season_months = {
        "Winter": {11, 12, 1, 2},
        "Spring": {3, 4, 5},
        "Summer": {6, 7, 8},
        "Fall": {9, 10}
    }

    planting_month_match = df_temp.apply(
        lambda x: x["planting_date"].month in season_months.get(x["season"], set()),
        axis=1
    )
    report["planting_month_out_of_season_count"] = int((~planting_month_match).sum())

    print("Planting month out of expected season:", report["planting_month_out_of_season_count"])

    # ==================================================
    # 8. NUMERIC RANGE CHECKS
    # ==================================================
    print("\n[8] NUMERIC RANGE CHECKS")

    numeric_non_negative_cols = [
        "area_size", "quantity_sold", "unit_price", "total_sales", "total_expenses",
        "fertilizer_used", "water_used", "actual_yield", "distance_to_market",
        "transport_cost", "market_price", "rainfall", "humidity", "wind_speed",
        "demand_score", "supply_score"
    ]

    for col in numeric_non_negative_cols:
        report[f"{col}_negative_count"] = int((df[col] < 0).sum())
        print(f"{col} negative values:", report[f"{col}_negative_count"])

    report["area_size_too_large_count"] = int((df["area_size"] > 120).sum())
    report["temperature_out_of_range_count"] = int(((df["temperature"] < 5) | (df["temperature"] > 50)).sum())
    report["humidity_out_of_range_count"] = int(((df["humidity"] < 10) | (df["humidity"] > 100)).sum())
    report["rainfall_out_of_range_count"] = int(((df["rainfall"] < 0) | (df["rainfall"] > 150)).sum())
    report["wind_speed_out_of_range_count"] = int(((df["wind_speed"] < 0) | (df["wind_speed"] > 35)).sum())
    report["demand_score_out_of_range_count"] = int(((df["demand_score"] < 0) | (df["demand_score"] > 1)).sum())
    report["supply_score_out_of_range_count"] = int(((df["supply_score"] < 0) | (df["supply_score"] > 1)).sum())

    print("Area too large:", report["area_size_too_large_count"])
    print("Temperature out of range:", report["temperature_out_of_range_count"])
    print("Humidity out of range:", report["humidity_out_of_range_count"])
    print("Rainfall out of range:", report["rainfall_out_of_range_count"])
    print("Wind speed out of range:", report["wind_speed_out_of_range_count"])
    print("Demand score out of range:", report["demand_score_out_of_range_count"])
    print("Supply score out of range:", report["supply_score_out_of_range_count"])

    # ==================================================
    # 9. YIELD / SALES CONSISTENCY
    # ==================================================
    print("\n[9] YIELD / SALES CONSISTENCY")

    sales_ratio = df["quantity_sold"] / (df["actual_yield"] + 1e-6)
    yield_per_area = df["actual_yield"] / (df["area_size"] + 1e-6)

    min_possible = df["crop_name"].map(lambda c: crops_info[c]["yield_range"][0]) * 0.5
    max_possible = df["crop_name"].map(lambda c: crops_info[c]["yield_range"][1]) * 2.8

    report["quantity_sold_gt_actual_yield_count"] = int((df["quantity_sold"] > df["actual_yield"] + 1e-6).sum())
    report["quantity_sold_equal_zero_count"] = int((df["quantity_sold"] <= 0).sum())
    report["actual_yield_equal_zero_count"] = int((df["actual_yield"] <= 0).sum())
    report["sales_ratio_gt_1_count"] = int((sales_ratio > 1.0).sum())
    report["sales_ratio_lt_0_75_count"] = int((sales_ratio < 0.75).sum())
    report["sales_ratio_gt_0_999_count"] = int((sales_ratio > 0.999).sum())
    report["yield_per_area_too_low_count"] = int((yield_per_area < min_possible).sum())
    report["yield_per_area_too_high_count"] = int((yield_per_area > max_possible).sum())
    report["yield_per_area_zero_count"] = int((yield_per_area <= 0).sum())

    print("Quantity sold > actual yield:", report["quantity_sold_gt_actual_yield_count"])
    print("Quantity sold <= 0:", report["quantity_sold_equal_zero_count"])
    print("Actual yield <= 0:", report["actual_yield_equal_zero_count"])
    print("Sales ratio > 1:", report["sales_ratio_gt_1_count"])
    print("Sales ratio < 0.75:", report["sales_ratio_lt_0_75_count"])
    print("Sales ratio > 0.999:", report["sales_ratio_gt_0_999_count"])
    print("Yield too low:", report["yield_per_area_too_low_count"])
    print("Yield too high:", report["yield_per_area_too_high_count"])
    print("Yield per area <= 0:", report["yield_per_area_zero_count"])

    # ==================================================
    # 10. PRICE VALIDATION
    # ==================================================
    print("\n[10] PRICE VALIDATION")

    base_price_series = df["crop_name"].map(lambda c: crops_info[c]["base_price"])
    price_ratio = df["market_price"] / (df["unit_price"] + 1e-6)

    report["unit_price_non_positive_count"] = int((df["unit_price"] <= 0).sum())
    report["market_price_non_positive_count"] = int((df["market_price"] <= 0).sum())
    report["unit_price_too_low_vs_base_count"] = int((df["unit_price"] < base_price_series * 0.6).sum())
    report["unit_price_too_high_vs_base_count"] = int((df["unit_price"] > base_price_series * 4.0).sum())
    report["market_price_too_far_from_unit_price_count"] = int(((price_ratio < 0.9) | (price_ratio > 1.15)).sum())
    report["unit_price_gt_market_price_count"] = int((df["unit_price"] > df["market_price"] * 1.05).sum())

    print("Unit price <= 0:", report["unit_price_non_positive_count"])
    print("Market price <= 0:", report["market_price_non_positive_count"])
    print("Unit price too low:", report["unit_price_too_low_vs_base_count"])
    print("Unit price too high:", report["unit_price_too_high_vs_base_count"])
    print("Market vs unit price mismatch:", report["market_price_too_far_from_unit_price_count"])
    print("Unit price unexpectedly > market price:", report["unit_price_gt_market_price_count"])

    # ==================================================
    # 11. FINANCIAL CONSISTENCY
    # ==================================================
    print("\n[11] FINANCIAL CONSISTENCY")

    calc_profit = (df["total_sales"] - df["total_expenses"]).round(2)
    calc_total_sales = (df["quantity_sold"] * df["unit_price"]).round(2)

    profit_margin_sales = df["profit"] / (df["total_sales"] + 1e-6)
    expense_sales_ratio = df["total_expenses"] / (df["total_sales"] + 1e-6)

    report["profit_formula_mismatch_count"] = int(((df["profit"] - calc_profit).abs() > 1.0).sum())
    report["total_sales_formula_mismatch_count"] = int(((df["total_sales"] - calc_total_sales).abs() > 1.0).sum())
    report["profit_margin_gt_300pct_count"] = int((profit_margin_sales > 3.0).sum())
    report["profit_margin_lt_minus90pct_count"] = int((profit_margin_sales < -0.9).sum())
    report["expenses_gt_5x_sales_count"] = int((expense_sales_ratio > 5.0).sum())

    print("Profit formula mismatch:", report["profit_formula_mismatch_count"])
    print("Total sales formula mismatch:", report["total_sales_formula_mismatch_count"])
    print("Profit margin > 300%:", report["profit_margin_gt_300pct_count"])
    print("Profit margin < -90%:", report["profit_margin_lt_minus90pct_count"])
    print("Expenses > 5x sales:", report["expenses_gt_5x_sales_count"])

    # ==================================================
    # 12. RESOURCE USAGE CHECK
    # ==================================================
    print("\n[12] RESOURCE USAGE CHECK")

    fert_per_area = df["fertilizer_used"] / (df["area_size"] + 1e-6)
    water_per_area = df["water_used"] / (df["area_size"] + 1e-6)
    transport_per_km = df["transport_cost"] / (df["distance_to_market"] + 1e-6)

    report["fertilizer_per_area_too_low_count"] = int((fert_per_area < 0.15).sum())
    report["fertilizer_per_area_too_high_count"] = int((fert_per_area > 1.0).sum())
    report["water_per_area_too_low_count"] = int((water_per_area < 5).sum())
    report["water_per_area_too_high_count"] = int((water_per_area > 25).sum())
    report["transport_cost_zero_with_positive_distance_count"] = int(((df["distance_to_market"] > 0) & (df["transport_cost"] <= 0)).sum())
    report["transport_per_km_too_low_count"] = int((transport_per_km < 0.005).sum())
    report["transport_per_km_too_high_count"] = int((transport_per_km > 0.2).sum())

    print("Fertilizer too low:", report["fertilizer_per_area_too_low_count"])
    print("Fertilizer too high:", report["fertilizer_per_area_too_high_count"])
    print("Water too low:", report["water_per_area_too_low_count"])
    print("Water too high:", report["water_per_area_too_high_count"])
    print("Transport cost zero with positive distance:", report["transport_cost_zero_with_positive_distance_count"])
    print("Transport per km too low:", report["transport_per_km_too_low_count"])
    print("Transport per km too high:", report["transport_per_km_too_high_count"])

    desert_avg = df[df["region_type"] == "Desert"]["water_used"].mean()
    rural_avg = df[df["region_type"] == "Rural"]["water_used"].mean()
    report["desert_avg_water_used"] = float(desert_avg)
    report["rural_avg_water_used"] = float(rural_avg)
    report["desert_water_not_higher_than_rural_flag"] = bool(desert_avg <= rural_avg)

    print("Average water desert:", report["desert_avg_water_used"])
    print("Average water rural:", report["rural_avg_water_used"])
    print("Desert water <= rural water flag:", report["desert_water_not_higher_than_rural_flag"])

    # ==================================================
    # 13. WEATHER LOGIC CHECK
    # ==================================================
    print("\n[13] WEATHER LOGIC CHECK")

    desert_temp = df[df["region_type"] == "Desert"]["temperature"].mean()
    rural_temp = df[df["region_type"] == "Rural"]["temperature"].mean()
    desert_rain = df[df["region_type"] == "Desert"]["rainfall"].mean()
    rural_rain = df[df["region_type"] == "Rural"]["rainfall"].mean()

    report["average_desert_temperature"] = float(desert_temp)
    report["average_rural_temperature"] = float(rural_temp)
    report["average_desert_rainfall"] = float(desert_rain)
    report["average_rural_rainfall"] = float(rural_rain)
    report["desert_not_hotter_than_rural_flag"] = bool(desert_temp <= rural_temp)
    report["desert_not_drier_than_rural_flag"] = bool(desert_rain >= rural_rain)

    print("Average desert temperature:", report["average_desert_temperature"])
    print("Average rural temperature:", report["average_rural_temperature"])
    print("Average desert rainfall:", report["average_desert_rainfall"])
    print("Average rural rainfall:", report["average_rural_rainfall"])
    print("Desert not hotter than rural flag:", report["desert_not_hotter_than_rural_flag"])
    print("Desert not drier than rural flag:", report["desert_not_drier_than_rural_flag"])

    # ==================================================
    # 14. PEST LOGIC
    # ==================================================
    print("\n[14] PEST LOGIC")

    high_risk = (df["humidity"] > 60) | (df["rainfall"] > 25) | (df["temperature"] > 34)

    pest_high = df.loc[high_risk, "pest_indicator"].mean()
    pest_low = df.loc[~high_risk, "pest_indicator"].mean()

    report["pest_rate_high_risk"] = float(pest_high)
    report["pest_rate_low_risk"] = float(pest_low)
    report["pest_not_higher_in_high_risk_flag"] = bool(pest_high <= pest_low)

    print("Pest rate (high risk weather):", report["pest_rate_high_risk"])
    print("Pest rate (low risk weather):", report["pest_rate_low_risk"])
    print("Pest not higher in high-risk weather flag:", report["pest_not_higher_in_high_risk_flag"])

    # ==================================================
    # 15. DEMAND / SUPPLY FEATURE CHECK
    # ==================================================
    print("\n[15] DEMAND / SUPPLY FEATURE CHECK")

    demand_low_max = df.loc[df["demand_level"] == "Low", "demand_score"].max() if (df["demand_level"] == "Low").any() else np.nan
    demand_high_min = df.loc[df["demand_level"] == "High", "demand_score"].min() if (df["demand_level"] == "High").any() else np.nan

    supply_low_max = df.loc[df["supply_level"] == "Low", "supply_score"].max() if (df["supply_level"] == "Low").any() else np.nan
    supply_high_min = df.loc[df["supply_level"] == "High", "supply_score"].min() if (df["supply_level"] == "High").any() else np.nan

    report["demand_score_low_level_mean"] = float(df.loc[df["demand_level"] == "Low", "demand_score"].mean())
    report["demand_score_medium_level_mean"] = float(df.loc[df["demand_level"] == "Medium", "demand_score"].mean())
    report["demand_score_high_level_mean"] = float(df.loc[df["demand_level"] == "High", "demand_score"].mean())

    report["supply_score_low_level_mean"] = float(df.loc[df["supply_level"] == "Low", "supply_score"].mean())
    report["supply_score_medium_level_mean"] = float(df.loc[df["supply_level"] == "Medium", "supply_score"].mean())
    report["supply_score_high_level_mean"] = float(df.loc[df["supply_level"] == "High", "supply_score"].mean())

    report["demand_level_order_issue_flag"] = bool(
        not (
            report["demand_score_low_level_mean"] <
            report["demand_score_medium_level_mean"] <
            report["demand_score_high_level_mean"]
        )
    )

    report["supply_level_order_issue_flag"] = bool(
        not (
            report["supply_score_low_level_mean"] <
            report["supply_score_medium_level_mean"] <
            report["supply_score_high_level_mean"]
        )
    )

    print("Demand mean score by level:",
          report["demand_score_low_level_mean"],
          report["demand_score_medium_level_mean"],
          report["demand_score_high_level_mean"])
    print("Supply mean score by level:",
          report["supply_score_low_level_mean"],
          report["supply_score_medium_level_mean"],
          report["supply_score_high_level_mean"])
    print("Demand level ordering issue:", report["demand_level_order_issue_flag"])
    print("Supply level ordering issue:", report["supply_level_order_issue_flag"])

    # ==================================================
    # 16. SUFFICIENCY CONSISTENCY
    # ==================================================
    print("\n[16] SUFFICIENCY CONSISTENCY")

    adjusted_supply_score = np.clip(0.92 * df["supply_score"] + 0.08 * df["demand_score"], 0, 1)
    ratio = np.round(adjusted_supply_score / (df["demand_score"] + 1e-4), 6)

    eps = 1e-3

    shortage_wrong = ((df["sufficiency_status"] == "Shortage") & (ratio >= 0.92 + eps)).sum()
    balanced_wrong = ((df["sufficiency_status"] == "Balanced") & ((ratio < 0.92 - eps) | (ratio > 1.08 + eps))).sum()
    surplus_wrong = ((df["sufficiency_status"] == "Surplus") & (ratio <= 1.08 - eps)).sum()

    report["sufficiency_shortage_rule_violation_count"] = int(shortage_wrong)
    report["sufficiency_balanced_rule_violation_count"] = int(balanced_wrong)
    report["sufficiency_surplus_rule_violation_count"] = int(surplus_wrong)
    report["sufficiency_total_rule_violation_count"] = int(shortage_wrong + balanced_wrong + surplus_wrong)

    print("Shortage rule violations:", report["sufficiency_shortage_rule_violation_count"])
    print("Balanced rule violations:", report["sufficiency_balanced_rule_violation_count"])
    print("Surplus rule violations:", report["sufficiency_surplus_rule_violation_count"])
    print("Total sufficiency rule violations:", report["sufficiency_total_rule_violation_count"])

    # ==================================================
    # 17. BOOLEAN CHECK
    # ==================================================
    print("\n[17] BOOLEAN CHECK")

    report["invalid_pest_indicator_type_count"] = int((~df["pest_indicator"].isin([True, False])).sum())

    print("Invalid pest_indicator values:", report["invalid_pest_indicator_type_count"])

    # ==================================================
    # 18. DISTRIBUTION / IMBALANCE CHECK
    # ==================================================
    print("\n[18] DISTRIBUTION / IMBALANCE CHECK")

    crop_distribution = df["crop_name"].value_counts(normalize=True)
    demand_distribution = df["demand_level"].value_counts(normalize=True)
    supply_distribution = df["supply_level"].value_counts(normalize=True)
    suff_distribution = df["sufficiency_status"].value_counts(normalize=True)
    customer_distribution = df["customer_type"].value_counts(normalize=True)
    region_distribution = df["region_type"].value_counts(normalize=True)

    report["crop_max_share_pct"] = float(crop_distribution.max() * 100)
    report["crop_min_share_pct"] = float(crop_distribution.min() * 100)
    report["demand_distribution_pct"] = (demand_distribution * 100).round(2).to_dict()
    report["supply_distribution_pct"] = (supply_distribution * 100).round(2).to_dict()
    report["sufficiency_distribution_pct"] = (suff_distribution * 100).round(2).to_dict()
    report["customer_distribution_pct"] = (customer_distribution * 100).round(2).to_dict()
    report["region_distribution_pct"] = (region_distribution * 100).round(2).to_dict()

    print("Max crop share %:", round(report["crop_max_share_pct"], 2))
    print("Min crop share %:", round(report["crop_min_share_pct"], 2))
    print("Demand distribution %:", report["demand_distribution_pct"])
    print("Supply distribution %:", report["supply_distribution_pct"])
    print("Sufficiency distribution %:", report["sufficiency_distribution_pct"])
    print("Customer distribution %:", report["customer_distribution_pct"])
    print("Region distribution %:", report["region_distribution_pct"])

    # ==================================================
    # 19. GROUP-LEVEL BUSINESS CHECKS
    # ==================================================
    print("\n[19] GROUP-LEVEL BUSINESS CHECKS")

    profit_by_crop = df.groupby("crop_name")["profit"].mean()
    profit_by_region = df.groupby("region_type")["profit"].mean()
    sales_by_customer = df.groupby("customer_type")["total_sales"].mean()

    report["lowest_avg_profit_crop"] = str(profit_by_crop.idxmin())
    report["highest_avg_profit_crop"] = str(profit_by_crop.idxmax())
    report["lowest_avg_profit_region"] = str(profit_by_region.idxmin())
    report["highest_avg_profit_region"] = str(profit_by_region.idxmax())
    report["highest_avg_sales_customer_type"] = str(sales_by_customer.idxmax())

    print("Lowest avg profit crop:", report["lowest_avg_profit_crop"])
    print("Highest avg profit crop:", report["highest_avg_profit_crop"])
    print("Lowest avg profit region:", report["lowest_avg_profit_region"])
    print("Highest avg profit region:", report["highest_avg_profit_region"])
    print("Highest avg sales customer type:", report["highest_avg_sales_customer_type"])

    # ==================================================
    # 20. CORRELATION SANITY CHECKS
    # ==================================================
    print("\n[20] CORRELATION SANITY CHECKS")

    numeric_cols = [
        "area_size", "quantity_sold", "unit_price", "total_sales", "total_expenses",
        "fertilizer_used", "water_used", "actual_yield", "profit", "distance_to_market",
        "transport_cost", "market_price", "temperature", "humidity",
        "rainfall", "wind_speed", "demand_score", "supply_score"
    ]

    corr = df[numeric_cols].corr(numeric_only=True)

    report["corr_area_actual_yield"] = float(corr.loc["area_size", "actual_yield"])
    report["corr_quantity_total_sales"] = float(corr.loc["quantity_sold", "total_sales"])
    report["corr_distance_transport_cost"] = float(corr.loc["distance_to_market", "transport_cost"])
    report["corr_water_area"] = float(corr.loc["water_used", "area_size"])
    report["corr_market_price_demand_score"] = float(corr.loc["market_price", "demand_score"])
    report["corr_actual_yield_supply_score"] = float(corr.loc["actual_yield", "supply_score"])

    print("Correlation area_size vs actual_yield:", round(report["corr_area_actual_yield"], 3))
    print("Correlation quantity_sold vs total_sales:", round(report["corr_quantity_total_sales"], 3))
    print("Correlation distance_to_market vs transport_cost:", round(report["corr_distance_transport_cost"], 3))
    print("Correlation water_used vs area_size:", round(report["corr_water_area"], 3))
    print("Correlation market_price vs demand_score:", round(report["corr_market_price_demand_score"], 3))
    print("Correlation actual_yield vs supply_score:", round(report["corr_actual_yield_supply_score"], 3))

    # ==================================================
    # 21. OUTLIER CHECK (IQR)
    # ==================================================
    print("\n[21] OUTLIER CHECK (IQR)")

    outlier_cols = [
        "area_size", "actual_yield", "quantity_sold", "unit_price",
        "total_sales", "total_expenses", "profit", "water_used",
        "fertilizer_used", "demand_score", "supply_score"
    ]

    for col in outlier_cols:
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr = q3 - q1
        low = q1 - 1.5 * iqr
        high = q3 + 1.5 * iqr
        outlier_count = int(((df[col] < low) | (df[col] > high)).sum())
        report[f"{col}_outlier_count_iqr"] = outlier_count
        print(f"{col} outliers (IQR):", outlier_count)

    # ==================================================
    # 22. PROFIT DISTRIBUTION
    # ==================================================
    print("\n[22] PROFIT DISTRIBUTION")

    report["positive_profit_count"] = int((df["profit"] > 0).sum())
    report["negative_profit_count"] = int((df["profit"] < 0).sum())
    report["zero_profit_count"] = int((df["profit"] == 0).sum())

    report["positive_profit_pct"] = float(round((df["profit"] > 0).mean() * 100, 2))
    report["negative_profit_pct"] = float(round((df["profit"] < 0).mean() * 100, 2))
    report["zero_profit_pct"] = float(round((df["profit"] == 0).mean() * 100, 4))
    report["avg_profit"] = float(df["profit"].mean())
    report["median_profit"] = float(df["profit"].median())

    print("Positive Profit Count:", report["positive_profit_count"])
    print("Negative Profit Count:", report["negative_profit_count"])
    print("Zero Profit Count:", report["zero_profit_count"])
    print("Positive Profit %:", report["positive_profit_pct"])
    print("Negative Profit %:", report["negative_profit_pct"])
    print("Zero Profit %:", report["zero_profit_pct"])
    print("Average Profit:", round(report["avg_profit"], 2))
    print("Median Profit:", round(report["median_profit"], 2))

    # ==================================================
    # 23. FINAL WARNING SUMMARY
    # ==================================================
    print("\n[23] FINAL WARNING SUMMARY")

    warning_flags = {
        "has_missing_values": report["missing_values_total"] > 0,
        "has_duplicate_rows": report["duplicate_rows"] > 0,
        "has_invalid_categories": (
            report["invalid_crop_name_count"] > 0 or
            report["invalid_season_count"] > 0 or
            report["invalid_market_location_count"] > 0 or
            report["invalid_farm_location_count"] > 0 or
            report["invalid_region_type_count"] > 0 or
            report["invalid_customer_type_count"] > 0 or
            report["invalid_demand_level_count"] > 0 or
            report["invalid_supply_level_count"] > 0 or
            report["invalid_sufficiency_status_count"] > 0
        ),
        "has_bad_date_order": report["bad_date_order_count"] > 0,
        "has_sales_yield_issue": report["quantity_sold_gt_actual_yield_count"] > 0,
        "has_score_range_issue": (
            report["demand_score_out_of_range_count"] > 0 or
            report["supply_score_out_of_range_count"] > 0
        ),
        "has_price_issues": (
            report["unit_price_non_positive_count"] > 0 or
            report["market_price_non_positive_count"] > 0 or
            report["unit_price_too_low_vs_base_count"] > 0 or
            report["unit_price_too_high_vs_base_count"] > 0
        ),
        "has_profit_formula_issues": report["profit_formula_mismatch_count"] > 0,
        "has_sufficiency_rule_issues": report["sufficiency_total_rule_violation_count"] > 0,
        "demand_level_order_issue": report["demand_level_order_issue_flag"],
        "supply_level_order_issue": report["supply_level_order_issue_flag"],
        "desert_temp_logic_issue": report["desert_not_hotter_than_rural_flag"],
        "desert_water_logic_issue": report["desert_water_not_higher_than_rural_flag"],
        "pest_logic_issue": report["pest_not_higher_in_high_risk_flag"]
    }

    report["warning_flags"] = warning_flags

    for k, v in warning_flags.items():
        print(f"{k}: {v}")

    print("\n==========================================")
    print(" END OF ADVANCED DATASET CHECK REPORT ")
    print("==========================================")

    return report

report = check_dataset_realism_advanced(df)
```
